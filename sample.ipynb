```python
# --- Класс 1-го уровня (Базовый) ---
class TV_Base:

    def __init__(self, firm, diagonal, power):

        self.firm = firm
        self.diagonal = diagonal
        self.power = power

    def calculate_Q(self):

        return self.diagonal + (0.05 * self.power)

    def display_info(self):

        print(f"--- Телевизор (Базовый) ---")
        print(f"Фирма: {self.firm}")
        print(f"Диагональ: {self.diagonal} дюймов")
        print(f"Мощность звука: {self.power} дБ")
        print(f"Качество (Q): {self.calculate_Q():.2f}")
        print("-" * 28)


# --- Класс 2-го уровня (Наследник) ---
class TV_Extended(TV_Base):

    def __init__(self, firm, diagonal, power, country):

        super().__init__(firm, diagonal, power)
        self.country = country

    def calculate_Q(self):

        base_q = super().calculate_Q()

        if self.country.lower() == "япония":
            return 2 * base_q
        elif self.country.lower() in ["сингапур", "корея"]:
            return 1.5 * base_q
        else:
            return base_q

    def display_info(self):

        print(f"--- Телевизор (Расширенный) ---")
        print(f"Фирма: {self.firm}")
        print(f"Страна-производитель: {self.country}")
        print(f"Диагональ: {self.diagonal} дюймов")
        print(f"Мощность звука: {self.power} дБ")
        print(f"Качество (Qp): {self.calculate_Q():.2f}")
        print("-" * 28)


if __name__ == "__main__":
    print("Демонстрация работы классов:\n")

    tv1 = TV_Base(firm="Samsung", diagonal=32, power=20)
    tv1.display_info()


    tv_japan = TV_Extended(firm="Sony", diagonal=55, power=30, country="Япония")
    tv_japan.display_info()

    tv_singapore = TV_Extended(firm="LG", diagonal=43, power=25, country="Сингапур")
    tv_singapore.display_info()

    tv_korea = TV_Extended(firm="LG", diagonal=50, power=40, country="Корея")
    tv_korea.display_info()

    tv_china = TV_Extended(firm="Xiaomi", diagonal=65, power=50, country="Китай")
    tv_china.display_info()
```
